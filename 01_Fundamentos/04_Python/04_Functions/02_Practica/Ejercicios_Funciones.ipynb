## Ejercicio 1
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana. La función constará de un único argumento numérico y una salida de tipo string

````python
def dias_semana(nindice):
    dias = ["Lunes", "Martes", "Miercoles", "Jueves", "Viernes", "Sabado", "Domingo"]

    if (nindice <= 0):
        return "Error de entrada"

    if (nindice > 7):
        return (dias[(nindice - 7) - 1]) 
    else:
        return (dias[nindice - 1])

dia_input = int(input("Te voy a convertir ese numerin en un dia de la semana"))
print(dias_semana(dia_input))
````

## Ejercicio 3
Escibe una función que compare dos números. La función tiene dos argumentos y hay tres salidas posibles: que sean iguales, que el primero se  mayor que el segundo, o que el segundo sea mayor que el primero

````python
num1_e3 = int(input("Dime un numero"))
num2_e3 = int(input("Dime otro numero"))

def compare(num1, num2):
    if (num1 == num2):
        return "Iguales"
    if (num1 > num2):
        return "Primero mayor"
    if (num2 > num1):
        return "Segundo mayor"

print("El resultado de la comporativa es: " + compare(num1_e3, num2_e3))
````

## Ejercicio 4
Escribe una función que sea un contador de letras. En el primer argumento tienes que introducir un texto, y el segundo que sea la letra a contar. La función tiene que devolver un entero con el número de veces que aparece esa letra, tanto mayuscula, como minúscula

````python
def letras_count(texto: str, letra: str):
    return int(texto.count(letra.lower()))

texto_input_e4 = input("Dame una frase")
letra_input_e4 = input("Ahora dime que letra de esa frase quieres que te cuente")

print('La letra "' + letra_input_e4 + '" aparece ' + str(letras_count(texto_input_e4, letra_input_e4)) + " veces.")
````

## Ejercicio 5
Escribe una función que tenga un único argumento, un string. La salida de la función tiene que ser un diccionario con el conteo de todas las letras de ese string.

````python
def diccionario_letras(texto: str):
    lista_car = list()
    dicc_car = dict()

    for i in range(len(texto)):
        lista2_car = list()
        lista2_car.append(texto[i])
        lista2_car.append(str(texto.count(texto[i])))

        lista_car.insert(i, lista2_car)
        dicc_car[lista2_car[0]] = lista2_car[1]

    print(dicc_car)


input_e5 = input("Dime una frase y te cuento las letras")
diccionario_letras(input_e5)
````

## Ejercicio 6
Escribir una función que añada o elimine elementos en una lista. La función necesita los siguientes argumentos:
* lista: la lista donde se añadirán o eliminarán los elementos
* comando: "add" o "remove"
* elemento: Por defecto es None.

````python
lista_e6 = ["Epicness", "Awesome", "You are a star"]

def list_modif(lista: list, command: str, elemento=None):

    if (command.lower() == "add"):
        lista.append(elemento)

    if (command.lower() == "remove"):
        lista.remove(elemento)

    return lista

lista_e6 = list_modif(lista_e6, "remoVe", "Awesome")
print(lista_e6)
````

## Ejercicio 7
Crea una función que reciba un número arbitrario de palabras, y devuelva una frase completa, separando las palabras con espacios.

````python
def crea_frase(*palabra: str):
    frase = ""
    for i in palabra:
        if (palabra.index(i) != (len(palabra)) - 1):
            frase = frase + i + " "
        else:
            frase = frase + i + ""
    return frase

print(crea_frase("Hola","Que","Tal", "Estamos", "MINECRAFT"))
````

## Ejercicio 8
Escribe un programa que obtenga el enésimo número de la [serie de Fibonacci](https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci). Tienes que crear una función recursiva con un único argumento.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

````python
num_fib = 0
list_fib = []

def fibonacci(numero):

    if (len(list_fib) < 1):
        list_fib.append(0)
        list_fib.append(1)

    elif (numero > 0):
        list_fib.append(list_fib[-1] + list_fib[-2])

    if (numero >= 0):
        fibonacci(numero - 1)

    return list_fib[-1]

print(fibonacci(20))
````

## Ejercicio 9
Define en una única celda las siguientes funciones:
* Función que calcule el área de un cuadrado
* Función que calcule el area de un triángulo
* Función que calcule el área de un círculo

En otra celda, calcular el area de:
* Dos círculos de radio 10 + un triángulo de base 3 y altura 7
* Un cuadrado de lado = 10 + 3 círculos (uno de radio = 4 y los otros dos de radio = 6) + 5 triángulos de base = 2 + altura = 4

````python
import math
def sq_area(lado):
    return lado ** 2

def tri_area(base, altura):
    return (base * altura) / 2

def circ_area(radio):
    return math.pi * (radio ** 2)
````

````python
for i in range(1,3):
    print("Circulo " + str(i) + ": ",end="")
    print(circ_area(10))

print("Triangulo 1:", tri_area(3,7))
print("Cuadradro 1:", sq_area(10))
print("Circulo 3:", circ_area(4))

for i in range(4,6):
    print("Circulo " + str(i) +": ",end="")
    print(circ_area(6))

for i in range(2,7):
    print("Triangulo "+ str(i) + ": ",end="")
    print(tri_area(2,4))
````
