## Ejercicio 1
1. Elimina todos los VW de la siguiente lista. Crea una lista nueva para ello.
2. Hazlo de otra manera. Crea una lista nueva, y ve añadiendo todos los que no sean VW
3. En vez de `for`, usa un bucle `while`
> ejer_1 = ["VW", "Audi", "Renault", "VW", "BMW"]

```python
ejer_1 = ["VW", "Audi", "Renault", "VW", "BMW"]

lista2_e1 = ejer_1.copy()

for items in ejer_1:
    if (lista2_e1.count("VW") != 0):
        lista2_e1.remove("VW")
print("Lista2 = " + str(lista2_e1)) # 1 

lista3_e1 = list()

for items in range(len(ejer_1)):
    if (ejer_1[items] != "VW"):
        lista3_e1.append(ejer_1[items])
print("Lista3 = " + str(lista3_e1)) # 2

lista4_e1 = list()

contador = 0
while contador < len(ejer_1):
    if (ejer_1[contador] != "VW"):
        lista4_e1.append(ejer_1[contador])
    contador = contador + 1
print("Lista4 = " + str(lista4_e1))
```

## Ejercicio 3
1. Escribe un programa que vaya pidiendo numeros al usuario. Cuando el usuario introduzca el 0, el programa tiene que imprimir por pantalla el sumatorio de todos los numeros positivos introducidos
2. Además de la suma, queremos que imprima también una lista con todos los números.

```python
num_e3 = int(input("Dime un numero"))

lista_e3 = list()

while num_e3 != 0:
    lista_e3.append(num_e3)
    num_e3 = int(input("Dime un numero"))

sum_lista_e3 = 0

for item_e3 in range(len(lista_e3)):
    if lista_e3[item_e3] > 0:
        sum_lista_e3 = sum_lista_e3 + lista_e3[item_e3]
    
print("Tus numeros: ", lista_e3)
print("El sumatorio de los numeros positivos es:", sum_lista_e3)
```

## Ejercicio 11
Suponiendo que siempre se cumple que las dos listas del ejercicio son iguales de tamaño, implementa un programa que calcule el ratio ingreso/gasto de cada elemento, y lo guarde en una nueva lista. Si hubiese algún problema con los datos, hay que almacenar un mensaje de error en la lista del resultado. Usa `try/except`
> ingresos = [100, 200, 500, 100, 600]
>
> gastos = [50, 20, 70, 0, 25]

```python
ingresos_e11 = [100, 200, 500, 100, 600]
gastos_e11 = [50, 20, 70, 0, 25]

ratios = list()
if (len(ingresos_e11) == len(gastos_e11)):
    for i in range(len(ingresos_e11)):
        try:
            calculo_e11 = ingresos_e11[i] / gastos_e11[i]
            ratios.append(calculo_e11)
        except:
            print("Error con datos encontrado: Ingresos =", ingresos_e11[i],"y Gastos =",gastos_e11[i])
            ratios.append("ERROR")

    for item in ratios:
        print(item, end= " ")
else:
    print("Conjunto de datos incorrectos. Ingresos y Gastos deben tener la misma cantidad de datos")


        
```

## Ejercicio 12
Muy parecido al anterior, aunque ahora tendremos que excepcionar más tipos de errores con estos nuevos datos. En este caso, almacena un mensaje diferente en la lista resultado, para cada error.

> ingresos = [100, "HHH", 500, 100, 600, 900, 150]
>
> gastos = [50, 20, 70, 0, 25]

```python
ingresos_e12 = [100, "HHH", 500, 100, 600, 900, 150]
gastos_e12 = [50, 20, 70, 0, 25,30,25]

ratios = list()
if (len(ingresos_e12) == len(gastos_e12)):
    for i in range(len(ingresos_e12)):
        try:
            calculo_e12 = ingresos_e12[i] / gastos_e12[i]
            ratios.append(calculo_e12)
        except ZeroDivisionError:
            print("Error con datos encontrado: Ingresos =", ingresos_e12[i],"y Gastos =",gastos_e12[i])
            ratios.append("ERROR_ZERO")
        except TypeError:
            print("Error con datos encontrado: El tipo de ingresos y gastos no coinciden. (",ingresos_e12[i],") y (", gastos_e12[i],")")
            ratios.append("ERROR_TIPOS")

    for item in ratios:
        print(item, end= " ")
        
else:
    print("Conjunto de datos incorrectos. Ingresos y Gastos deben tener la misma cantidad de datos")
```
