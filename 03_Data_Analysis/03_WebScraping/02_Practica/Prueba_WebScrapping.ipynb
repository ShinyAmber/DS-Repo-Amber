# Prueba WebScrapping

```python
import time
import requests
from bs4 import BeautifulSoup as bs
import pandas as pd

# Configuramos Pandas para que muestre textos largos sin cortarlos
pd.set_option('max_colwidth', 800)

# Definimos nuestro 'disfraz' de navegador (User-Agent) para evitar bloqueos 403
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/124.0 Safari/537.36"
}
```

```python
url = "https://www.apple.com/es/iphone-18-pro/specs/"
response = requests.get(url, headers=headers, timeout=10)
response.raise_for_status()

soup = bs(response.content, "html.parser")

specs = soup.find_all("div", class_="techspecs")
print(f"Total de secciones: {len(specs)}")

primer_spec = specs[0]
```

```python
modelo = primer_spec.find("span", class_="header-iphone-pro-max").get_text()


```

```python
capacidad_minima = primer_spec.find("li", role="listitem").get_text()
```

```python
peso = primer_spec.find("div", class_="weight-wrapper pro-max")
peso = peso.find("p", class_="weight-copy").get_text()
```

```python
anchura = primer_spec.find("div", class_="techspecs-section section-size-and-weight")
anchura = anchura.find_all("div", class_="techspecs-column small-spans-2-columns")[-1]
anchura = anchura.find("figcaption", class_="diagram-text diagram-text-width")
anchura = anchura.find("p").get_text()
```

```python
altura = primer_spec.find("div", class_="techspecs-section section-size-and-weight")
altura = altura.find_all("div", class_="techspecs-column small-spans-2-columns")[-1]
altura = altura.find("figcaption", class_="diagram-text diagram-text-height")
altura = altura.find("p").get_text()
```

```python
print(f"Modelo: {modelo}")
print(f"Capacidad Minima: {capacidad_minima}")
print(f"Peso: {peso}")
print(f"Anchura: {anchura}")
print(f"Altura: {altura}")
```
