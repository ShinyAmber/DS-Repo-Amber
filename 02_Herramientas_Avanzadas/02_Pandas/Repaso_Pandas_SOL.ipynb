#### 2. Selección y Filtros (`loc` vs `iloc`)
El indexado en Pandas puede ser confuso si intentamos usar los corchetes estándar de las listas de Python. Para evitar ambigüedades, utilizamos los indexadores especiales.

*(Nota: Recuérda la metáfora visual: `loc` es leer lo que pone en la "etiqueta/pegatina" de la caja, mientras que `iloc` es contar la posición física de la caja empezando desde cero, independientemente de lo que ponga en su pegatina.)*

```python
# Uso de iloc: Selección implícita por posición 
# (Filas de la 0 a la 4, primeras 3 columnas)

subset_iloc = planets.iloc[0:5, 0:3]
display(subset_iloc)
```

```python
# Uso de loc: Selección explícita por etiquetas y máscaras booleanas
# Buscamos planetas descubiertos por 'Radial Velocity' después del año 2010
condicion = (planets['method'] == 'Radial Velocity') & (planets['year'] > 2010)

# Aplica la máscara sobre las columnas "method, distance, year" usando .loc
subset_loc = planets.loc[condicion, ['method', 'distance', 'year']]
display(subset_loc.head())
```

#### 3. Manejo de Valores Nulos (NaN)
En el mundo real, los datos rara vez están completos. Pandas utiliza `NaN` ("Not a Number") como su forma de señalar datos faltantes.

*(Nota: La aparición de un NaN no es un error de código, sino la forma que tiene Pandas de avisar que falta información en ese registro. Aquí pueden decidir si rellenar los datos (`fillna`) o eliminarlos (`dropna`).)*

```python
# Comprobamos cuántos valores nulos hay por columna
planets.isnull().sum()
```

```python
# Estrategia A: Eliminar las filas que contengan cualquier valor nulo con dropna()
# Crea una variable llamada planets_limpio y agrégale el df sin nulos

planets_limpio = planets.dropna()

print(f"\nTamaño original: {planets.shape}. \
      Tamaño sin nulos: {planets_limpio.shape}")
```

```python
# Estrategia B: Rellenar los valores nulos 
# (por ejemplo, con la media de la columna o un 0)

# fill_value=0 es un salvavidas que Pandas hereda para gestionar la alineación 
#       de índices con datos faltantes

planets['mass'] = planets['mass'].fillna(planets['mass'].mean())
```

#### 4. Agregaciones y GroupBy
El concepto de `groupby` en Pandas se entiende mejor a través de la filosofía *Split, Apply, Combine* (Dividir, Aplicar, Combinar).
*   **Split**: Rompe y agrupa el DataFrame dependiendo del valor de la clave que especifiques.
*   **Apply**: Calcula alguna función (agregación, transformación o filtro) exclusivamente dentro de cada uno de esos grupos individuales.
*   **Combine**: Fusiona los resultados en un único array o DataFrame de salida.

Una vez que el método ``groupby`` ha ejecutado el paso de Split (Dividir), llega el momento de ``Apply`` (Aplicar) la lógica sobre cada grupo aislado. Pandas nos ofrece estas cuatro herramientas que se diferencian de la siguiente manera:

*   **Aggregation (Agregación):** Es una operación matemática de reducción. Toma todas las filas de un grupo, calcula un valor estadístico (como .sum() o .mean()) y devuelve una única fila resumen por cada grupo.

*   **Filter (Filtrado):** Es un proceso de descarte a nivel colectivo. Evalúa una condición estadística conjunta y devuelve las filas originales, pero borra por completo los grupos enteros que no cumplan con dicha condición.

*   **Transform (Transformación):** Es una operación de alteración que respeta la estructura. Devuelve las mismas filas originales, pero con sus valores matemáticamente alterados basándose en los cálculos de su propio grupo (por ejemplo, para centrar datos restando la media grupal).

*   **Apply (Aplicar):** Es la navaja suiza de Pandas. Le pasas el bloque de datos del grupo entero a una función completamente arbitraria donde puedes hacer lo que quieras con él, y devuelve un número escalar, una nueva Serie o un DataFrame distinto.


*(Nota: Hacer `df.groupby('key')` por sí solo no calcula nada, sino que devuelve un objeto `DataFrameGroupBy` especial con evaluación perezosa (lazy evaluation), que solo actúa cuando se le aplica una función.)*

```python
# Agregación básica: Calcular la media del período orbital 
#       para cada método de descubrimiento

# Permite seleccionar una columna específica después de hacer el groupby 
#       y antes de aplicar la agregación, lo cual es mucho más eficiente


media_orbital = planets.groupby('method')['orbital_period'].mean()
display(media_orbital)
```

```python
# Uso de filter: Una operación de filtrado permite descartar datos 
#       basándose en las propiedades conjuntas del grupo

# Filtramos para quedarnos solo con los métodos 
#       que hayan descubierto más de 30 planetas


metodos_frecuentes = planets.groupby('method').filter(lambda x: x['number'].count() > 30)

display(metodos_frecuentes['method'].value_counts())
```

```python
# Uso de transform con lambda: Una transformación devuelve una versión alterada
#        de los datos completos manteniendo exactamente la misma forma (shape) 
#               que la entrada original


# Centramos el año de descubrimiento restando la media del grupo
# Una lambda es simplemente una función anónima de usar y tirar


planets['year_centered'] = planets.groupby('method')['year'].transform(lambda x: x - x.mean())
```

#### 5. Unir y Combinar Tablas
Como Data Scientists, constantemente tendremos que integrar datos de múltiples fuentes.

*   ¿Quieres simplemente pegar bloques de datos (arriba/abajo o lado a lado)?<br> 
    -> Usa `pd.concat()`, que es un pegamento estructural. No busca relaciones lógicas entre los datos, solo une estructuras.
<br>

*   ¿Quieres cruzar datos buscando coincidencias en una columna?<br>
    -> Usa `pd.merge()`, que es la unión relacional.
<br>

*   ¿Quieres cruzar datos buscando coincidencias en los índices de las filas?<br>
-> Usa ``df.join()``, que es el hermano cómodo de merge. Es un método de conveniencia diseñado y optimizado para cruzar DataFrames basándose en sus índices (las etiquetas de las filas) en lugar de en sus columnas.

##### Concat

```python
# Creamos dos mini-DataFrames ficticios sobre investigadores 
#       para demostrar las uniones


df_investigadores = pd.DataFrame({
    'method': ['Radial Velocity', 'Transit', 'Imaging'],
    'investigador_principal': ['Dr. Smith', 'Dra. García', 'Dr. Lee']
})

df_presupuestos = pd.DataFrame({
    'method': ['Transit', 'Imaging', 'Astrometry'],
    'presupuesto_M': [5.2, 1.8, 0.9]
})
```

```python
# 1. pd.concat(): Pega las tablas una debajo de la otra (axis=0 por defecto)
# Funciona como un pegamento estructural

unidos_vertical = pd.concat([df_investigadores, df_presupuestos], ignore_index=True)

print("Resultado de pd.concat (Estructural):")
display(unidos_vertical)
```

##### Merge

```python
# 2. pd.merge(): Combina tablas basándose en el valor de una columna en común(llave)
# 'how="inner"' solo conserva las filas donde la llave existe en ambas tablas

cruce_relacional = pd.merge(df_investigadores, df_presupuestos, on='method', how='inner')

print("\nResultado de pd.merge (Relacional SQL-style):")
display(cruce_relacional)
```

```python
# Integración con nuestro dataset principal
# Añadimos el investigador principal a nuestro dataset de planetas 
#       basándonos en el método usando left

planets_merged = pd.merge(planets, df_investigadores, on='method', how='left')
display(planets_merged.head())
```

##### Join

```python
# 1. Convertimos la columna compartida ('method') en el índice de ambas tablas 
#       para que el método join() pueda cruzarlas correctamente
investigadores_idx = df_investigadores.set_index('method')
presupuestos_idx = df_presupuestos.set_index('method')

display(investigadores_idx)
display(presupuestos_idx)
print("--------------------------------------")

# 2. Aplicamos el join (por defecto hace un 'left' join) 
#       y guardalo en resultado_join

resultado_join = investigadores_idx.join(presupuestos_idx)

print("Resultado de df.join (Cruzando por índices):")
display(resultado_join)

# ---------------------------------------------------------
# ALTERNATIVA (Unión bruta por posición de fila):
# Si quisieras pegarlas por su número de fila original sin importar el contenido, 
# tendrías que usar lsuffix y rsuffix para evitar el error por la columna duplicada:
# df_investigadores.join(df_presupuestos, lsuffix='_inv', rsuffix='_presup')
```
